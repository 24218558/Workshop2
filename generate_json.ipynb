#### Generate JSON

```python
def loadVideosToInfile(folder):
    videos = os.listdir(folder)
    with open('inputshortclip', 'w') as inputfile:
        for v in videos:
            if v.split('.')[-1] == 'mp4':
                inputfile.write(v+'\n')
```
